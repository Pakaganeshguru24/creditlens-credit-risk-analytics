```python
df.loc[
    df["loan_percent_income"] <= 0,
    ["person_income", "loan_amnt", "loan_percent_income", "loan_status"]
]
```

```python
df.loc[
    df["loan_percent_income"] <= 0,
    "calculated_loan_percent_income"
] = (
    df["loan_amnt"] / df["person_income"]
)

df.loc[
    df["loan_percent_income"] <= 0,
    ["loan_amnt", "person_income",
     "loan_percent_income", "calculated_loan_percent_income"]
]
```

```python
(df["loan_percent_income"] <= 0).sum()
```

```python
mask = df["loan_percent_income"] == 0

df.loc[mask, "loan_percent_income"] = (
    df.loc[mask, "loan_amnt"] /
    df.loc[mask, "person_income"]
)
```
